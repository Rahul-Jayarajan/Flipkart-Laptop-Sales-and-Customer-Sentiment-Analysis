#### Business Features 

```python
# Create a Column for categorize Sentiment Poor (0–2), Average (2–3), Good (3–4), and Excellent (4–5).
labels=['Poor', 'Average', 'Good', 'Excellent']
df["rating_category"] = pd.cut(df["rating"],bins = [0,2,3,4,5], labels=labels )
df
```

## Extracting

#### Extracting Brand Name

```python
# Extract the brand_Name


brands = ['Apple','ASUS','Lenovo','HP','Acer','MSI','Dell']
def extract_brand(product):
    for brand in brands:
        if brand.lower() in product.lower():
            return brand
    return 'Other'

df['brand'] = df['product_name'].apply(extract_brand)
```

#### Extracting Hardware Specs

```python
#Extract Hardware Specs

#RAM 
df['ram_gb'] = df['product_name'].str.extract(r'(\d+)\s*GB')

#Storage
df['storage_gb'] = df['product_name'].str.extract(r'/(\d+)\s*GB SSD')


#Extract GPU

def extract_gpu(text):

    text = str(text)

    if 'RTX 4050' in text:
        return 'RTX 4050'

    elif 'RTX 4060' in text:
        return 'RTX 4060'

    elif 'RTX 3050' in text:
        return 'RTX 3050'

    elif 'RTX 2050' in text:
        return 'RTX 2050'

    else:
        return 'Integrated'

df['gpu'] = df['product_name'].apply(extract_gpu)


```

#### Extracting High Value Features

```python
#Battery Mention 


#If you want to check whether a column contains a string, use .str.contains():
battery_keywords = [
    'battery',
    'backup',
    'charge',
    'charging',
    'drain'
]

df['battery_mentioned'] = (
    df['review']
    .str.lower()
    .str.contains('battery|backup|charge|charging|drain', na=False)  # use '|'.join(battery_keywords) instead 'battery|backup|charge|charging|drain'
)
```

```python
# Performance Mention
performance_keywords = [
    'performance',
    'fast',
    'speed',
    'smooth',
    'gaming',
    'lag'
]
df['performance_mentioned'] = (
    df['review']
    .str.lower()
    .str.contains('|'.join(performance_keywords), na=False)
)
```

```python
# Display Mention
display_keywords = [
    'display',
    'screen',
    'brightness'
]

df['display_mentioned'] = (
    df['review']
    .str.lower()
    .str.contains('|'.join(display_keywords), na=False)
)
```

```python
#Heating Issue
heating_keywords = [
    'heat',
    'heating',
    'hot',
    'thermal'
]

df['heating_issue'] = (
    df['review']
    .str.lower()
    .str.contains('|'.join(heating_keywords), na=False)
)
```

## IDENTIFY GAMING LAPTOP

```python
#Identify Gaming Laptops
gaming_keywords = [
    'gaming',
    'tuf',
    'rog',
    'nitro',
    'victus',
    'legion',
    'loq'
]

def gaming(product):
    for game in gaming_keywords:
        if game.lower() in product.lower():
            return game
    return "NO" 

df['is_gaming'] = df['product_name'].apply(gaming)
```
